### random seed 고정

```python
import numpy as np
import tensorflow as tf
import random as python_random 
import os

SEED = 42

np.random.seed(SEED)
python_random.seed(SEED)
tf.random.set_seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)
os.environ['TF_DETERMINISTIC_OPS'] = '1'
os.environ['TF_CUDNN_DETERMINISTIC'] = '1'
os.environ['TF_CUDNN_USE_FRONTEND '] = '1'

tf.__version__
```

### data 불러오기

```python
# 모든 행 출력
# pd.set_option('display.max_rows', None)

df = pd.read_csv('./forNewGraph.csv', index_col=0)
df
```

### 그래프 그리기

```python
# 2x3 그리드에 산점도 그리기
fig, axes = plt.subplots(1, 3, figsize=(15, 5),
                         sharex=True, sharey=True)

# Matplotlib의 기본 스타일 설정 변경
plt.rcParams['font.size'] = 13  # 전체 글꼴 크기
plt.rcParams['axes.labelsize'] = 10  # 축 레이블 크기
# plt.rcParams['xtick.labelsize'] = 10  # x축 눈금 레이블 크기
# plt.rcParams['ytick.labelsize'] = 10  # y축 눈금 레이블 크기

modals = ['weather','weather+soiling','weather+soiling+sunlight']
colors= ['dodgerblue', 'limegreen', 'orange']

for i, modal in enumerate(modals):

    lr = LinearRegression()
    lr.fit(df['Power generation (kW)'].values.reshape(-1, 1), df[modal].values.reshape(-1, 1))
    r_squared = lr.score(df['Power generation (kW)'].values.reshape(-1, 1), df[modal].values.reshape(-1, 1))

    axes[i].grid(visible=True, linestyle='--',  axis='y')
    axes[i].grid(visible=True, linestyle='--',  axis='x')
    
    
    axes[i].scatter(df['Power generation (kW)'].values.reshape(-1, 1), df[modal].values.reshape(-1, 1), color=colors[i], alpha=0.7)
    axes[i].plot(df['Power generation (kW)'].values.reshape(-1, 1), lr.predict(df['Power generation (kW)'].values.reshape(-1, 1)), color='red', label='선형 회귀 직선')
    axes[i].plot(df['Power generation (kW)'].values.reshape(-1, 1), df['Power generation (kW)'].values.reshape(-1, 1), color='black', label='ideal', linestyle=':')
    axes[i].set_title(modal)
    #axes[i].set_xlabel('Actual Pow_gen (kWh)', fontsize=15)
    #axes[i].set_ylabel('Forecasted Pow_gen (kWh)', fontsize=15)
#     axes[i].set_ylim(0, 350)
fig.tight_layout()
```
