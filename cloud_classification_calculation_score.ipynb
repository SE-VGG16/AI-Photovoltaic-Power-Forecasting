```python
import numpy as np
import tensorflow as tf
import random as python_random
import os

SEED = 42

np.random.seed(SEED)
python_random.seed(SEED)
tf.random.set_seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)
os.environ['TF_DETERMINISTIC_OPS'] = '1'
os.environ['TF_CUDNN_DETERMINISTIC'] = '1'
os.environ['TF_CUDNN_USE_FRONTEND '] = '1'

tf.__version__
```

```python
model = tf.keras.models.load_model('./best_model.h5')
```

```python
dataset_folder ="/app/MultiModal/dataset/image/20230605_ch1_crop/test/Image/"
```

```python
import pandas as pd 

class_names = ['A-sky', 'B-pattern', 'C-thick-dark', 'D-thick-white', 'E-veil']

df = pd.DataFrame(columns=["name"]+class_names)
df
```

```python
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
import os 
import warnings

warnings.filterwarnings(action='ignore')

img_height, img_width = 125, 125


for f in os.listdir(dataset_folder):
    img = tf.keras.utils.load_img(os.path.join(dataset_folder,f), target_size=(img_height, img_width))

    img_array = tf.keras.utils.img_to_array(img)
    img_array = tf.expand_dims(img_array, 0) # Create a batch

    predictions = model.predict(img_array)
    score = tf.nn.softmax(predictions[0])

    class_pred = tf.math.argmax(score)
    
    row = {"name":f}
    for i in range(5):
        row[class_names[i]] = score[i].numpy()*100 
        
    df= df.append(row, ignore_index=True)
```
